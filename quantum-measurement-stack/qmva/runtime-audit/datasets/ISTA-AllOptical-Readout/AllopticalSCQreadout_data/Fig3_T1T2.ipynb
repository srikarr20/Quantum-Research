```python
data_exp = np.load('data_ramsey_exp.npy')
data_fits = np.load('data_ramsey_fits.npy')
```

```python
colors = ['#66318F', '#75ba77', '#c43b3b', '#e36d7a']
colors_light = ['#9c61c9', '#9EC49F', '#E97777', '#e36d7a']
f, ax1 = plt.subplots(1,1, figsize=(4,3))



ms = 6
mew = 0.4
lw = 1

ax1.plot(data_fits[0][0]/1000, data_fits[0][1], color=colors[0], lw=lw, alpha=0.6, label='MW -> MW')
ax1.plot(data_fits[1][0]/1000, data_fits[1][1], color=colors[1], lw=lw, alpha=0.8, label='MW -> Opt')
ax1.plot(data_fits[2][0]/1000, data_fits[2][1], color=colors[2], lw=lw, alpha=0.6, label='Opt -> Opt')

ax1.plot(data_exp[0][0]/1000, data_exp[0][1], '.', mec=colors_light[0], ms=ms, mew = mew, mfc=(0, 0, 0, 0.0))
ax1.plot(data_exp[1][0]/1000, data_exp[1][1], '.', mec=colors_light[1], ms=ms, mew = mew, mfc=(0, 0, 0, 0.0))
ax1.plot(data_exp[2][0]/1000, data_exp[2][1], '.', mec=colors_light[2], ms=ms, mew = mew, mfc=(0, 0, 0, 0.0))
ax1.plot(data_exp[0][0]/1000, data_exp[0][1], '.', mec=colors[0], ms=ms, mew = 0, mfc=colors[0], alpha=0.2)
ax1.plot(data_exp[1][0]/1000, data_exp[1][1], '.', mec=colors[1], ms=ms, mew = 0, mfc=colors[1], alpha=0.2)
ax1.plot(data_exp[2][0]/1000, data_exp[2][1], '.', mec=colors[2], ms=ms, mew = 0, mfc=colors[2], alpha=0.2)



ax1.grid(linewidth = 0.3, alpha=0.5)

color = '#242424'
ax1.xaxis.label.set_color(color)
ax1.tick_params(axis='x', colors=color, direction="in", width=0.5, length=4)
ax1.tick_params(axis='x', which='minor', color=color, width=0.5, direction='in', length=2)
ax1.yaxis.label.set_color(color)
ax1.tick_params(axis='y', colors=color, direction="in", width=0.5, length=4)
ax1.tick_params(axis='y', which='minor', color=color, width=0.5, direction='in', length=2)
plt.tick_params(axis="both", which="both", bottom=True, top=False,    
                labelbottom=True, left=True, right=False, labelleft=True, width=0.5)

# plt.legend()

ax1.set_xlim([-0.0,4.1])
ax1.set_ylim([0.0,0.95])
ax1.xaxis.set_major_locator(MultipleLocator(1))
ax1.xaxis.set_major_formatter(FormatStrFormatter('%d'))
ax1.xaxis.set_minor_locator(MultipleLocator(0.5))

ax1.yaxis.set_major_locator(MultipleLocator(0.2))
ax1.yaxis.set_major_formatter(FormatStrFormatter('%.1f'))
ax1.yaxis.set_minor_locator(MultipleLocator(0.1))

plt.show()
```
